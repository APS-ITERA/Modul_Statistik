### **Soal 1**

```python
sampel_ch4 = [0.42, 0.55, 7.21, 0.38, 0.49, 9.50, 0.61, 5.80, 0.45]

# (a) Menyaring nilai konsentrasi metana yang merupakan lonjakan (> 1.0 ppb)
spikes = [x for x in sampel_ch4 if x > 1.0]
data_tanpa_spike = [x for x in sampel_ch4 if x <= 1.0]

# (b) Hitung nilai rata-rata
mean_dengan_spike = sum(sampel_ch4) / len(sampel_ch4)
mean_tanpa_spike = sum(data_tanpa_spike) / len(data_tanpa_spike)

print(f"Nilai lonjakan (spike): {spikes}")
print(f"Rata-rata dengan spike: {mean_dengan_spike:.2f} ppb")
print(f"Rata-rata tanpa spike: {mean_tanpa_spike:.2f} ppb")
```

Bagian (a): Menggunakan list comprehension untuk memisahkan data ke dalam dua list. spikes mengambil nilai di atas 1.0 ppb, sementara data_tanpa_spike menyimpan data normal (<= 1.0 ppb).

Bagian (b): Rata-rata dihitung dengan membagi total jumlah elemen dengan panjang (jumlah) elemen list. Rata-rata seluruh data adalah 2.82 ppb, sedangkan jika nilai spike dikeluarkan, rata-rata konsentrasi metana normal hanya 0.48 ppb.

### **Soal 2**

```python
import numpy as np

# Kolom: [Ketinggian (km), Suhu (K), Tekanan (hPa)]
data_atmosfer = np.array([
    [2.0, 275.15, 795.0],
    [8.0, 236.00, 356.0],
    [12.0, 216.65, 193.0],
    [18.0, 216.65, 75.0],
    [25.0, 221.55, 25.0],
    [32.0, 228.65, 8.8]
])

# (a) Menyaring baris untuk daerah Tropopause/Stratosfer Bawah
# Syarat: 10.0 <= Ketinggian <= 30.0 km DAN Suhu < 225.0 K
mask = (data_atmosfer[:, 0] >= 10.0) & (data_atmosfer[:, 0] <= 30.0) & (data_atmosfer[:, 1] < 225.0)
data_tersaring = data_atmosfer[mask]

# (b) Mengekstrak hanya nilai Tekanan (hPa)
tekanan_tersaring = data_tersaring[:, 2]

print("Data Tersaring:\n", data_tersaring)
print("Nilai Tekanan saja (hPa):", tekanan_tersaring)
```

Bagian (a): Kondisi filter (masking) digabungkan dengan operator bitwise &. Kolom pertama [:, 0] (ketinggian) dibatasi dari 10.0 km hingga 30.0 km, dan kolom kedua [:, 1] (suhu) dibatasi di bawah 225.0 K.

Bagian (b): Dari baris data yang lolos kondisi (baris ke-3, ke-4, dan ke-5), nilai ditarik hanya pada indeks kolom terakhir yaitu [:, 2] yang mewakili Tekanan (hPa). Output yang dihasilkan adalah array [193.0, 75.0, 25.0].

### **Soal 3**

(a)Formulasi Model Matematika Linear Programming:

Variabel Keputusan:  
* $x_1$ = Jumlah peluncuran Balon Tipe Radiosonde   

* $x_2$ = Jumlah peluncuran Balon Tipe Ozonsonde  

Fungsi Tujuan (Minimumkan Biaya):
* Minimize$$Z = 2.000.000 x_1 + 5.000.000 x_2$$

Kendala (Constraints):

* Sensor Suhu & Tekanan:$$3x_1 + 2x_2 \ge 12$$

* Profil Ozon:$$2x_2 \ge 6$$

* Batas Peluncuran Harian:$$x_1 + x_2 \le 8$$

* Non-negativitas:$$x_1 \ge 0, \quad x_2 \ge 0$$




(b) Script Python

```python
from scipy.optimize import linprog
import math

# Koefisien fungsi tujuan: meminimumkan biaya (dalam Rupiah)
c = [2000000, 5000000]

# Koefisien untuk sisi kiri kendala pertidaksamaan (A_ub)
# Kendala 1: -3x1 - 2x2 <= -12  (dari 3x1 + 2x2 >= 12)
# Kendala 2:  0x1 - 2x2 <= -6   (dari 2x2 >= 6)
# Kendala 3:   1x1 + 1x2 <= 8   (dari x1 + x2 <= 8)
A_ub = [
    [-3, -2],
    [0, -2],
    [1, 1]
]

# Koefisien batas sisi kanan (b_ub)
b_ub = [-12, -6, 8]

# Batas variabel keputusan x1 dan x2 (keduanya >= 0)
x_bounds = (0, None)

# Optimisasi model
solusi = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=[x_bounds, x_bounds], method='highs')

print(f"Status solusi: {solusi.message}")
print(f"Radiosonde (x1) optimal: {math.ceil(solusi.x[0])} peluncuran")
print(f"Ozonsonde (x2) optimal: {math.ceil(solusi.x[1])} peluncuran")
print(f"Biaya Minimum: Rp {int(solusi.fun)}")
```

(a) Alasan Algoritma Non-Deterministik Lebih Cocok:
Metode deterministik (seperti Gradient Descent) membutuhkan kalkulasi turunan dari sebuah fungsi matematika atau bentang medan yang bersifat kontinu. Masalah rute penjelajahan antar-titik (TSP) dan non-kontinuitas topografi Mars menyebabkan fungsi tujuannya (kalkulasi jarak antar simpul) memiliki lompatan, sehingga turunannya (gradien) tidak terdefinisi. Algoritma Genetic Algorithm (GA) menggunakan pendekatan probabilistik (teori seleksi alam secara acak-terarah) untuk mengeksplorasi ruang solusi, sehingga model pencariannya bebas-turunan (derivative-free) dan menghindari kelemahan terjebak di suatu posisi optimum lokal karena medan berbatuan ekstrim.



(b) Proses Algoritma Genetika untuk Kasus TSP Rover Mars:
Untuk rute 6 titik ($P_0$ hingga $P_5$), mekanisme GA beroperasi secara sekuensial:   
1. Membangkitkan rute-rute (kromosom) secara acak. Dalam representasi TSP, setiap rute merupakan kromosom dengan urutan permutasi (contoh: P0-P3-P1-P5-P4-P2-P0).
2. Menghitung total jarak setiap kromosom berdasarkan topografi kawah yang non-linear. Nilai "fitness" berbanding terbalik dengan jarak total tempuh (semakin pendek total metrik medannya, semakin tinggi fitness-nya).
3. Memilih sepasang kromosom dengan fitness tertinggi untuk disilangkan menjadi populasi generasi berikutnya (menggunakan metode seleksi roda roulette atau tournament).
4. Memotong sebagian rute dari Kromosom Parent A lalu melengkapinya dengan urutan sisa dari Kromosom Parent B tanpa adanya node duplikat. (Dalam TSP sering dipakai variasi Order Crossover).
5. Mengubah susunan DNA rute secara acak untuk mempertahankan keragaman populasi, seperti menukar titik secara probabilitas (swap posisi P3 dengan P4 secara acak).
6. Populasi dievaluasi ulang. Generasi berputar kembali ke langkah ke-2 hingga ditemukannya rute jarak optimal yang konvergen/tidak berubah lagi setelah ribuan generasi.
